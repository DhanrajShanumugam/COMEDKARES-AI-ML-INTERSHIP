```python
df['director']=df['director'].fillna('No Director Specified')

df.head()
```

```python
directors_list = pd.DataFrame()
print(directors_list)
```

```python
directors_list = df['director'].str.split(',', expand=True).stack()
print(directors_list)
```

```python
directors_list = directors_list.to_frame()
directors_list
```

```python
directors_list.columns = ['director']
print(directors_list)
```

```python
directors = directors_list.groupby(['director']).size().reset_index(name='Total count')
print(directors)
```

```python
directors = directors[directors['director'] != "No Director specified"]
```

```python
print(directors)
```

```python
directors = directors.sort_values(by=['Total Count'], ascending = False)
print(directors)
```

```python
top4Directors = directors.head()
print(top4Directors)
```

```python
top5Directors = top5Directors.sort_values(by=['Total Count'])
barChart = px.bar(top5Directors, x='Total Count', y = 'Director', title = 'Top 5 Directors on Netflix')
barChart.show()
```

```python
df['cast']=df['cast'].fillna('No cast specified')
cast_df = pd.DataFrame()
cast_df = df['cast'].str.split(',',expand=True).stack()
cast_df = cast_df.to_frame()
cast_df.columns = ['Actor']
actors = cast_df.groupby(['Actor']).size().reset_index(name = 'Total Count')
actors = actors[actors.Actor != 'No cast specified']
actors = actors.sort_values(by=['Total Count'], ascending=False)
top4Actors = actors.head()
top4Actors = top4Actors.sort_values(by=['Total Count'])
barChart2 = px.bar(top4Actors, x='Total Count', y='Actor', title='Top 4 Actors on Netflix')
barChart2.show()
```

```python
df1 = df[['type', 'release_year']]
df1 = df1.rename(columns = {"release_year":"Release Year", "type": "Type"})
df2 = df1.groupby(['Release Year', 'Type']).size().reset_index(name='Total Count')
```

```python
df2 = df2[df2['Release Year']>=2015]
graph = px.line(df2, x = "Release Year", y="Total Count", color = "Type", title = "Trend of Content Produced on Netfilx Every Year")
graph.show()
```

```python
df3 = df[['release_year', 'description']]
df3 = df3.rename(columns = {'release_year':'Release Year', 'description':'Description'})
for index, row in df3.iterrows():
  d=row['Description']
  testimonial = TextBlob(d)
  p = testimonial.sentiment.polarity
  if p==0:
    sent = 'Neutral'
  elif p>0:
    sent = 'Positive'
  else:
    sent = 'Negative'
  df3.loc[[index, 2], 'Sentiment']=sent

df3 = df3.groupby(['Release Year', 'Sentiment']).size().reset_index(name = 'Total Count')

df3 = df3[df3['Release Year']>2010]
barGraph = px.bar(df3, x="Release Year", y="Total Count", color = "Sentiment", title = "Sentiment Analysis of Content on Netflix")
barGraph.show()
```
